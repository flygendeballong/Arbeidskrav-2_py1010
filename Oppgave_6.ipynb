```python
# Importering av NumPy for å gjøre matematiske beregninger
import numpy as np

# Importering av Matplotlib for å lage graf og lage plott (visualisere data)
import matplotlib.pyplot as plt

# Definering av funksjon
def fun_f(x): 
    ''' Funksjon som gir en graf
    x er inn-argumentet/parameteren
    '''
    return -x * 2 - 5

# Lager x-verdier med 200 punkter jevnt fordelt fra -10 til 10
x = np.linspace(-10, 10, 200)

# Beregning av tilhørende y-verdier
y = fun_f(x)

# Plotting av funksjonen
plt.plot(x, y)
plt.grid()
plt.xlabel('x-akse')
plt.ylabel('y-akse')
plt.title('f(x) = -x2 - 5')
```
