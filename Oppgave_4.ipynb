Andre program utvider dictionaryen med følgende informasjon: et land bruker velger fritt, samt hovedstaden og innbyggertallet i landet, som bruker også skriver inn. Av Victoria K.
```python
# Opprettelse av dictionaryen data
data = {
    'Norge': ['Oslo', 0.634], 
    'England': ['London', 8.982],  
    'Frankrike': ['Paris', 2.161], 
    'Italia': ['Roma', 2.873]
        }
```

```python
# Deklarering av variabelen for land (bruker-input)
land = input('Skriv inn et land')

print(data[land][0], 'er hovedstaden i', land, 'og det er', data[land][1], 'mill. innbyggere i', data[land][0])

```

```python
# Deklarering av variablene for nytt land, dette landets hovedstad og dette landets innbyggertall (bruker-input), samt utvidelse av dictionaryen data

nytt_land = input('Skriv inn et nytt land')

nytt_land_hovedstad = input('Skriv inn hovedstaden i landet du valgte')

nytt_land_innbyggere = float(input('Skriv inn antall innbyggere i landet du valgte'))

data[nytt_land]= [nytt_land_hovedstad, nytt_land_innbyggere]

print(data)

```
