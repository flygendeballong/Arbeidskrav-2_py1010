```python
# Importering av NumPy-pakken/biblioteket, da programmet skal gjøre matematiske beregninger/arbeide med tall
import numpy as np

# Deklarering av variabelen for grader
v_grad = float(input('Skriv inn gradtallet:' ))

# Deklarering av variabelen for radianer, ved uttrykket for omregning av grader til radianer
v_rad = v_grad*np.pi/180

# Utskriving av resultatet, med forklarende tekst
print(v_grad, 'grader er det samme som', round(v_rad,2), 'radianer')
```
