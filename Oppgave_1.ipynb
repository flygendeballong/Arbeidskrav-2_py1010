Program som angir hvor gammel brukeren blir i løpet av "nåværende år" 2024, av Victoria K. Csisar (victoria.kc@outlook.com). Oppdatert 13. november 2026

```python
# Deklarering av variabelen for alder
alder = int(input('Hvilket år er du født? ') )

# Deklarering av variabelen for nåværende år
nåværende_år = 2024

# Deklarering/overskriving av variabelen for alder, per nåværende år
alder = nåværende_år - alder

# Utskriving av beregnet alder i år, med forklarende tekst
print('Du blir', alder, 'i år')
```
