Program som regner ut areal og omkrets av en figur bestående av en rettvinklet trekant og en halvsirkel, av Victoria K. Csisar (victoria.kc@outlook.com). Oppdatert 13. november 2026.

```python
# Importering av NumPy-pakken for matematiske beregninger
import numpy as np

# Definering av funksjon
def fun_areal_og_omkrets(a, b):
    '''
    Beregner arealet og omkretsen av figuren.
    a og b er katetene til trekanten.
    '''
    # Deklarering av variabler for arealet og omkretsen av halvsirkelen, med formler for utregningen
    areal_halvsirkel = (np.pi*((a/2)**2))/2
    omkrets_halvsirkel = (2*np.pi*(a/2))/2

    # Deklarering av variabler for arealet av trekanten og omkretsen av "toppen av trekanten" (hypotenus + katet b). c er hypotenusen til trekanten
    areal_trekant = (a*b)/2
    c = np.sqrt((a**2)+(b**2))
    omkrets_toppenavtrekant = b + c

    # Deklarering av variabler for arealet og omkretsen av hele figuren
    totalt_areal = areal_halvsirkel + areal_trekant
    total_omkrets = omkrets_halvsirkel + omkrets_toppenavtrekant

    # Returnering av verdiene for areal og omkrets av hele figuren, for videre bruk i programmet
    return (totalt_areal, total_omkrets)
```

```python
# Deklarering av variabelen a (bruker-input)
a = float(input('Oppgi lengden på a: '))
```

```python
# Deklarering av variabelen b (bruker-input)
b = float(input('Oppgi lengden på b: '))
```

```python
# Kaller funksjonen med parameterne a og b, og lagrer resultatene
totalt_areal, total_omkrets = fun_areal_og_omkrets(a, b)

# Utskriving av beregnet areal og ytre omkrets med forklarende tekst
print('Arealet er', totalt_areal, 'og "ytre" omkrets er', total_omkrets)
```
