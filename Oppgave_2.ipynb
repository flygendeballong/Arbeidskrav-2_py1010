```python
# Deklarering av variabelen for antall elever
antall_elever = int(input('Skriv inn antall elever:' ))

# Importering av math-pakken/biblioteket, da vi trenger funksjonen for å runde opp desimaltall til nærmeste heltall
import math

# Deklarering av variabelen for antall pizzaer
antall_pizzaer = antall_elever * 1/4

print('Det må handles inn', math.ceil(antall_pizzaer), 'pizzaer til festen')
```
